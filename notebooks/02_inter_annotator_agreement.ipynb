# Inter-annotator agreement
Use a raw Label Studio JSON export containing independent annotations on the same tasks. Agreement must be computed before adjudication.

```python
import json
from pathlib import Path

from autolabeler.evaluation import cohen_kappa, krippendorff_alpha_nominal, label_studio_agreement, span_f1

export_path = Path("exports/retrieval.json")
if export_path.exists():
    report = label_studio_agreement(json.loads(export_path.read_text(encoding="utf-8")), control="relevance")
    report
else:
    print("Set export_path to a Label Studio JSON export.")
```

## Understand the measures
Cohen's κ handles two raters and corrects raw agreement for marginal label frequencies. Krippendorff's α accepts more raters and incomplete overlap. Both can become unstable for rare labels, so always publish raw agreement and prevalence. Span F1 measures character overlap and should accompany, not replace, categorical agreement.

```python
a = ["relevant", "relevant", "irrelevant", "uncertain"]
b = ["relevant", "irrelevant", "irrelevant", "uncertain"]
print(cohen_kappa(a, b))
print(
    krippendorff_alpha_nominal(
        {"t1": {"a": "relevant", "b": "relevant"}, "t2": {"a": "irrelevant", "b": "relevant", "c": "irrelevant"}}
    )
)
print(span_f1([(10, 25)], [(12, 25)]))
```
