# Label project analysis
Inspect completion, label prevalence, annotator load, lead time, and the size of the overlap set before interpreting system metrics.

```python
import json
from pathlib import Path

from autolabeler.analysis import analyze_export

export_path = Path("exports/retrieval.json")
if export_path.exists():
    export = json.loads(export_path.read_text(encoding="utf-8"))
    summary = analyze_export(export, control="relevance")
    summary
else:
    print("Set export_path to a Label Studio JSON export.")
```

```python
# Optional chart; install with: uv sync --extra notebooks
if export_path.exists():
    import matplotlib.pyplot as plt
    import pandas as pd

    pd.Series(summary["labels"]).sort_values().plot.barh(title="Label distribution")
    plt.xlabel("annotations")
    plt.tight_layout()
```
